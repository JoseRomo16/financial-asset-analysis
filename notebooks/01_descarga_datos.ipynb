El portafolio combina renta variable global, oro, criptomoneda, divisas (USD/COP, EUR/USD), el índice bursátil colombiano (COLCAP) y café Arabica como commodity exportador — clases de activos heterogéneas que permiten estudiar diversificación efectiva.

## Configuración inicial

Añadimos la raíz del proyecto al `sys.path` para poder importar desde `src/` sin instalar el paquete.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt

from src.data_loader import descargar_precios, resumen_datos, TICKERS_DEFAULT
from src.visualizer import graficar_precios

pd.set_option('display.float_format', '{:.4f}'.format)
```

## 1. Descarga de precios

Usamos los tickers por defecto del módulo y un rango de cinco años. La función guarda el archivo crudo en `data/raw/precios.csv` y el procesado (con forward fill) en `data/processed/precios.csv`.

```python
precios = descargar_precios(
    tickers=TICKERS_DEFAULT,
    inicio='2020-01-01',
    fin='2024-12-31',
    guardar=True,
)
precios.tail()
```

## 2. Diagnóstico de calidad: valores nulos

Las criptomonedas operan 24/7 mientras las acciones no, por lo que es esperable encontrar nulos en días no bursátiles antes del forward fill.

```python
precios.isna().sum()
```

## 3. Resumen del panel

Tabla con observaciones, rango temporal y estadísticos descriptivos de precio por activo.

```python
resumen_datos(precios)
```

## 4. Visualización rápida de precios

```python
graficar_precios(precios, guardar=False);
plt.show()
```
