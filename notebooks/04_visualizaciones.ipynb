# 04 — Dashboard de visualizaciones consolidadas

**Objetivo:** integrar las visualizaciones clave en una vista tipo dashboard, exportar todas las figuras a `reports/figures/` y producir la tabla resumen final.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt

from src.data_loader import cargar_precios_local
from src.metrics import (
    calcular_retornos,
    calcular_retorno_acumulado,
    calcular_volatilidad,
    calcular_drawdown,
    resumen_metricas,
)
from src.visualizer import (
    graficar_precios,
    graficar_retorno_acumulado,
    graficar_drawdown,
    graficar_correlacion,
    graficar_distribucion_retornos,
    graficar_retorno_interactivo,
    graficar_volatilidad_interactiva,
)

pd.set_option('display.float_format', '{:.4f}'.format)
```

## 1. Carga y preparación de datos

```python
precios = cargar_precios_local()
retornos = calcular_retornos(precios, metodo='aritmetico')
ret_acum = calcular_retorno_acumulado(retornos)
drawdown = calcular_drawdown(retornos)
vol = calcular_volatilidad(retornos, ventana=21, anualizar=True)

print(f'Periodo: {precios.index.min().date()} → {precios.index.max().date()}')
print(f'Activos analizados: {list(precios.columns)}')
```

## 2. Vista dashboard — gráficos clave

Combinamos las visualizaciones más relevantes para una lectura rápida del portafolio.

### 2.1 Precios normalizados

```python
graficar_precios(precios, guardar=True);
plt.show()
```

### 2.2 Retorno acumulado

```python
graficar_retorno_acumulado(ret_acum, guardar=True);
plt.show()
```

### 2.3 Drawdown

```python
graficar_drawdown(drawdown, guardar=True);
plt.show()
```

### 2.4 Distribución de retornos

```python
graficar_distribucion_retornos(retornos, guardar=True);
plt.show()
```

### 2.5 Correlaciones

```python
graficar_correlacion(retornos, guardar=True);
plt.show()
```

## 3. Versiones interactivas (HTML)

Las versiones Plotly se exportan como archivos HTML embebibles en cualquier reporte web.

```python
graficar_retorno_interactivo(ret_acum, guardar=True).show()
```

```python
graficar_volatilidad_interactiva(vol, guardar=True).show()
```
