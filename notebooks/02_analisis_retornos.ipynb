**Objetivo:** comparar trayectorias de los activos en términos relativos, calcular retornos diarios y acumulados, estudiar su distribución y la estructura de correlaciones.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt

from src.data_loader import cargar_precios_local
from src.metrics import (
    calcular_retornos,
    calcular_retorno_acumulado,
    resumen_metricas,
)
from src.visualizer import (
    graficar_precios,
    graficar_retorno_acumulado,
    graficar_retorno_interactivo,
    graficar_distribucion_retornos,
    graficar_correlacion,
)

pd.set_option('display.float_format', '{:.4f}'.format)
```

Bajas correlaciones ⇒ mejor diversificación. El triángulo inferior basta para visualizarla.
