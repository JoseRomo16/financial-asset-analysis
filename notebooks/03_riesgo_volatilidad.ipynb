# 03 — Riesgo y volatilidad

**Objetivo:** caracterizar el perfil de riesgo de cada activo: volatilidad rodante, drawdown, ratio de Sharpe y VaR histórico.

```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt

from src.data_loader import cargar_precios_local
from src.metrics import (
    calcular_retornos,
    calcular_volatilidad,
    calcular_drawdown,
    calcular_max_drawdown,
    calcular_sharpe,
    calcular_var,
)
from src.visualizer import (
    graficar_drawdown,
    graficar_volatilidad_interactiva,
)

pd.set_option('display.float_format', '{:.4f}'.format)
```

```python
precios = cargar_precios_local()
retornos = calcular_retornos(precios)
retornos.tail()
```

## 1. Volatilidad rodante (ventana 21 días, anualizada)

Un mes bursátil ≈ 21 días hábiles. Multiplicamos por √252 para expresarla en términos anuales y compararla con el rendimiento.

```python
vol = calcular_volatilidad(retornos, ventana=21, anualizar=True)
vol.tail()
```

## 2. Drawdown por activo

Visualizamos cuán lejos cae cada activo de su máximo histórico durante el periodo.

```python
drawdown = calcular_drawdown(retornos)
graficar_drawdown(drawdown, guardar=True);
plt.show()
```

### Máximo drawdown (peor caída desde un máximo)

```python
max_dd = calcular_max_drawdown(retornos).to_frame('max_drawdown')
max_dd
```

## 3. Ratio de Sharpe (anualizado, tasa libre de riesgo = 5%)

```python
sharpe = calcular_sharpe(retornos, tasa_libre_riesgo=0.05).to_frame('sharpe')
sharpe
```

## 4. Value at Risk histórico al 95%

Un VaR de 0.025 significa que con 95% de confianza la pérdida diaria no excede el 2.5%.

```python
var_95 = calcular_var(retornos, confianza=0.95).to_frame('var_95')
var_95
```

## 5. Volatilidad rodante interactiva

```python
fig = graficar_volatilidad_interactiva(vol, guardar=True)
fig.show()
```
